# Project 3: Zomato Restaurant Data Analysis

## Objective
Analyze Zomato restaurant data to find cuisine trends,
city-wise distribution, pricing patterns and impact of
online delivery on ratings.

## Key Findings
1. North Indian and Chinese are top cuisines in India
2. Delhi NCR, Noida and Gurgaon have most restaurants
3. Restaurants WITH online delivery have higher average ratings
4. Majority of restaurants fall in price range 1-2 (budget friendly)
5. Most restaurants rated between 3.0 and 4.0

## Business Recommendation
Zomato should push more restaurants to enable online delivery
as data shows it directly correlates with higher ratings
and better customer satisfaction.

## Output
5 visualizations showing cuisine trends, city distribution,
delivery impact on ratings and price range breakdown.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('data/zomato.csv',encoding='latin-1')

print(df.shape)
print(df.columns.tolist())
print(df.head())
print(df.isnull().sum())
print(df['Country Code'].value_counts())
```

```python
df = df[df['Country Code'] == 1]
```

```python
# Rating distribution
sns.histplot(df['Aggregate rating'],bins=20,color='coral')
plt.title('Restaurant rating distribution in india')
plt.savefig('charts/chart1_rating.png')
plt.show()
```

```python
# Top 10 cuisines
top_cuisines = df['Cuisines'].value_counts().head(10)
sns.barplot(x=top_cuisines.values,y=top_cuisines.index,palette="Blues_d")
plt.title("Top 10 cuisines in india")
plt.savefig('charts/chart2_cuisines.png')
plt.show()
```

```python
# Online delivery vs rating
sns.boxplot(x='Has Online delivery',y='Aggregate rating',data=df,palette='Set2')
plt.title('Online Delivery vs Restaurant Rating')
plt.savefig('charts/chart3_delivery.png')
plt.show()
```

```python
# Top 10 cities by restaurant count
top_cities = df['City'].value_counts().head(10)
sns.barplot(x=top_cities.values,y=top_cities.index,palette='Greens_d')
plt.title('Top 10 Cities by Restaurant Count')
plt.savefig('charts/chart4_cities.png')
plt.show()
```

```python
# Price range distribution
df['Price range'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title('Restaurant Price Range Distribution')
plt.savefig('charts/chart5_price.png')
plt.show()
```
